```python
client = OpenAI()

def call_api(messages, model=None, max_tokens=None, temperature=1.2):
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        n=1,
        max_tokens=max_tokens,
        response_format={'type': 'text'},
        temperature=temperature
    )
    return response

def generate_message(messages, system_message=None):
    # messages must be a list of tuples with (role (str), content (str))
    messages = []
    if system_message is None:
        system_message = {
            'role': 'system',
            'content': 'you are an intelligent AI assistant. you produce clear, straightforward answers, and if you are unsure, you say so'
            }
    messages.append(system_message)
    for role, content in messages:
        messages.append({
            'role': role,
            'content': content
        })
    return messages


# prompt = 'how many "r"s are there in strawberry?'
# messages = []
# system_message = {
#     'role': 'system',
#     'content': 'you are an intelligent AI assistant. you produce clear, straightforward answers, and if you are unsure, you say so'}
# prompt_message = {
#     'role': 'user',
#     'content': prompt
# }

# messages.append(system_message)
# messages.append(prompt_message)

response = call_api(messages=messages, model='gpt-4o')
```
